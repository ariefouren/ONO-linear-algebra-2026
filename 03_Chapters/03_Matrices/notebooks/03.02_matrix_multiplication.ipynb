## 1. כפל שורה בעמודה (הגדרה 3.4.1)

$$\begin{bmatrix}a_1&\cdots&a_n\end{bmatrix}\begin{bmatrix}b_1\\ \vdots\\ b_n\end{bmatrix}=\sum_{k=1}^{n}a_kb_k$$

בפייתון, כפל מטריצות נכתב עם האופרטור `@`. (הכוכבית `*` היא כפל **איבר-איבר** – לא כפל מטריצות!)

</div>

```python
row = np.array([[3, 1, 2]])          # 1x3
col = np.array([[1], [-1], [3]])     # 3x1
show("row @ col", row @ col)          # 1x1: 3 - 1 + 6 = 8
show("row * row (element-wise, NOT a matrix product)", row * row)
```

## 2. מכפלת מטריצות (הגדרה 3.4.2, דוגמה 3.8)

$[AB]_{ij}=\sum_k a_{ik}b_{kj}$, ו-$(m\times n)(n\times q)=m\times q$.

</div>

```python
A = np.array([[1, 2, -1],
              [0, 3,  2]])           # 2x3
B = np.array([[ 2, 1],
              [-1, 0],
              [ 4, 3]])              # 3x2
show("AB", A @ B)                     # 2x2
show("BA", B @ A)                     # 3x3

def entry(A, B, i, j):
    """[AB]_ij computed by the definition (i, j start at 1)."""
    return sum(A[i-1, k] * B[k, j-1] for k in range(A.shape[1]))
print("[AB]_12 by the definition:", entry(A, B, 1, 2))
```

<div dir="rtl">

כשהסדרים לא מתאימים – המכפלה אינה מוגדרת:

</div>

```python
C = np.ones((2, 2))
try:
    A @ C            # (2x3)(2x2)
except ValueError as e:
    print("ValueError:", e)
```

## 3. שורות ועמודות של מכפלה (למה 3.4.3)

$[AB]_i^r=[A]_i^rB$ ו-$[AB]_j^c=A[B]_j^c$.

</div>

```python
A = np.array([[1, 2], [3, 4], [5, 6]])
B = np.array([[1, 0, 2], [-1, 1, 0]])
print("row 2 of AB:   ", (A @ B)[1, :], "  =  [A]_2^r B:", A[1, :] @ B)
print("column 3 of AB:", (A @ B)[:, 2], "  =  A [B]_3^c:", A @ B[:, 2])
```

## 5. שחלוף של מכפלה (טענה 3.4.5, דוגמה 3.9)

$(AB)^t=B^tA^t$ – הסדר מתהפך.

</div>

```python
A = np.array([[1, 2, -1], [0, 3, 2]]); B = np.array([[2, 1], [-1, 0], [4, 3]])
show("(AB)^t", (A @ B).T)
show("B^t A^t", B.T @ A.T)
show("A^t B^t  (a 3x3 matrix - not even the same order!)", A.T @ B.T)
```

## 6. קיבוציות (משפט 3.5.1)

בודקים על מטריצות אקראיות: $(AB)C=A(BC)$.

</div>

```python
rng = np.random.default_rng(0)
A = rng.integers(-5, 6, (2, 3)); B = rng.integers(-5, 6, (3, 4)); C = rng.integers(-5, 6, (4, 2))
print(np.array_equal((A @ B) @ C, A @ (B @ C)))
```

## 7. הכפל אינו חילופי; מכפלה שהיא אפס (דוגמה 3.10)

</div>

```python
A = np.array([[1, 1], [0, 0]]); B = np.array([[1, 0], [1, 0]])
C = np.array([[1, 1], [1, 1]]); D = np.array([[1, -1], [-1, 1]])
show("AB", A @ B); show("BA", B @ A); show("CD", C @ D)

# how often do two random 2x2 integer matrices commute?
rng = np.random.default_rng(1)
trials = 10000
same = sum(np.array_equal(X @ Y, Y @ X)
           for X, Y in ((rng.integers(-3, 4, (2, 2)), rng.integers(-3, 4, (2, 2))) for _ in range(trials)))
print(f"commuting pairs: {same} out of {trials}")
```

## 8. מטריצת היחידה, פילוג, ונוסחת הכפל המקוצר (דוגמה 3.11)

</div>

```python
A = np.array([[1, 1], [0, 0]]); B = np.array([[1, 0], [1, 0]]); I = np.eye(2, dtype=int)
print("IA == A == AI:", np.array_equal(I @ A, A) and np.array_equal(A @ I, A))
S = A + B
show("(A+B)^2", S @ S)
show("A^2 + AB + BA + B^2", A@A + A@B + B@A + B@B)
show("A^2 + 2AB + B^2  (wrong in general)", A@A + 2*A@B + B@B)
```

## 9. יישום: גרפיקה ממוחשבת – הסדר קובע

נשמור צורה (בית קטן) כעמודות של מטריצה $P$ מסדר $2\times k$, ונפעיל עליה טרנספורמציות במכפלה אחת.

</div>

```python
P = np.array([[0, 2, 2, 1, 0, 0],
              [0, 0, 2, 3, 2, 0]], dtype=float)       # a house; last point closes the outline

R = np.array([[0, -1], [1, 0]])      # rotate 90 degrees counter-clockwise
S = np.array([[2, 0], [0, 1]])       # stretch x by 2

def draw(ax, Q, title, style="-o"):
    ax.plot(Q[0], Q[1], style); ax.set_title(title)
    ax.set_aspect("equal"); ax.grid(True); ax.set_xlim(-4, 5); ax.set_ylim(-3, 5)

fig, ax = plt.subplots(1, 3, figsize=(14, 4.5))
draw(ax[0], P, "P (original)")
draw(ax[1], S @ R @ P, "S R P: rotate, then stretch")
draw(ax[2], R @ S @ P, "R S P: stretch, then rotate")
plt.show()
show("SR", S @ R); show("RS", R @ S)
```

**נסו בעצמכם:** סיבוב בזווית כללית $\theta$ הוא $R_\theta=\begin{bmatrix}\cos\theta&-\sin\theta\\ \sin\theta&\cos\theta\end{bmatrix}$.

</div>

```python
import ipywidgets as widgets
@widgets.interact(theta=widgets.IntSlider(min=0, max=360, step=15, value=45),
                  sx=widgets.FloatSlider(min=0.5, max=2.5, step=0.25, value=1.0))
def transform(theta, sx):
    t = np.radians(theta)
    Rt = np.array([[np.cos(t), -np.sin(t)], [np.sin(t), np.cos(t)]])
    St = np.array([[sx, 0], [0, 1]])
    fig, ax = plt.subplots(figsize=(5, 5))
    ax.plot(P[0], P[1], "-o", alpha=.35, label="P")
    Q = St @ Rt @ P
    ax.plot(Q[0], Q[1], "-o", label="S R P")
    ax.set_aspect("equal"); ax.grid(True); ax.set_xlim(-6, 6); ax.set_ylim(-6, 6); ax.legend(); plt.show()
```

## 11. יישום: שכבה ברשת נוירונים

שכבה מחשבת $y=Wx+b$. אצווה של קלטים (עמודות של $X$) מחושבת במכפלה אחת $WX$.

</div>

```python
rng = np.random.default_rng(2)
W = rng.normal(size=(3, 4))          # 4 inputs -> 3 outputs
b = rng.normal(size=(3, 1))
X = rng.normal(size=(4, 5))          # a batch of 5 input vectors (columns)
Y = W @ X + b
show("Y = W X + b", Y)
print("same as one column at a time:", np.allclose(Y[:, [0]], W @ X[:, [0]] + b))
```

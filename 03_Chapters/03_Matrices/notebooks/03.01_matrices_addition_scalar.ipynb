```python
import numpy as np
import matplotlib.pyplot as plt
np.set_printoptions(suppress=True, precision=4)

def show(name, M):
    """Print a matrix with its name and its order (rows x columns)."""
    M = np.asarray(M)
    print(f"{name}  ({M.shape[0]}x{M.shape[1] if M.ndim > 1 else 1})")
    print(M, end="\n\n")
```

## 1. מטריצה, סדר ואיברים (דוגמה 3.1)

</div>

```python
A = np.array([[2, -1,  0,  4],
              [3,  5, -2,  1],
              [0,  7,  6, -3]])
show("A", A)
print("order:", A.shape)                        # (3, 4)
print("a_14 =", A[0, 3], " a_23 =", A[1, 2], " a_32 =", A[2, 1])
try:
    A[3, 0]                                     # a_41 does not exist
except IndexError as e:
    print("a_41:", e)
```

## 2. מטריצה הנתונה בנוסחה (דוגמה 3.2)

המטריצה $A=[a_{ij}]_{2\times 3}$ עם $a_{ij}=2i-j$. בקוד מוסיפים 1 לאינדקסים כדי לעבוד עם $i,j$ שמתחילים מ-1.

</div>

```python
A = np.fromfunction(lambda i, j: 2*(i+1) - (j+1), (2, 3), dtype=int)
show("A", A)

# try another formula, e.g. a_ij = i*j on a 3x3 matrix:
B = np.fromfunction(lambda i, j: (i+1)*(j+1), (3, 3), dtype=int)
show("B", B)
```

## 3. שוויון מטריצות (הגדרה 3.1.1)

שתי מטריצות שוות רק אם הן מאותו סדר **וגם** כל האיברים המתאימים שווים.

</div>

```python
P = np.array([[1, 2], [3, 4]])
Q = np.array([[1, 2], [3, 4]])
R = np.array([[1, 3], [2, 4]])                # same numbers, other places
print(np.array_equal(P, Q), np.array_equal(P, R))
print(np.array_equal(np.array([[1, 2]]), np.array([[1], [2]])))   # row vs column
```

## 4. שורות, עמודות ושחלוף (דוגמה 3.4)

</div>

```python
A = np.array([[1, 4, -2],
              [0, 3,  5]])
print("[A]_2^r =", A[1, :])
print("[A]_3^c =", A[:, 2].reshape(-1, 1).ravel(), "(as a column:)")
print(A[:, [2]])
show("A^t", A.T)
print("(A^t)^t == A ?", np.array_equal(A.T.T, A))    # claim 3.2.4
```

## 5. מטריצה סימטרית (הגדרה 3.2.6, דוגמה 3.5)

</div>

```python
def is_symmetric(M):
    M = np.asarray(M)
    return M.shape[0] == M.shape[1] and np.array_equal(M, M.T)

a, b = 3, -1
B = np.array([[1, a, 5],
              [3, 2, b],
              [5, -1, 0]])
print("symmetric with a=3, b=-1:", is_symmetric(B))
B[0, 1] = 4
print("after changing a:", is_symmetric(B))
```

## 6. חיבור, כפל בסקלר והפרש (דוגמה 3.6)

</div>

```python
A = np.array([[2, -1, 0], [1, 3, 4]])
B = np.array([[1,  2, -3], [0, -1, 2]])
show("A + B", A + B)
show("3A - 2B", 3*A - 2*B)
show("(A + B)^t", (A + B).T)
print("(A+B)^t == A^t + B^t ?", np.array_equal((A + B).T, A.T + B.T))   # theorem 3.3.7
```

## 7. פתרון משוואה מטריציונית (דוגמה 3.7)

$2X+A=5B \;\Longrightarrow\; X=\frac{1}{2}(5B-A)$

</div>

```python
A = np.array([[4, 0], [-2, 6]])
B = np.array([[2, 1], [0, 2]])
X = (5*B - A) / 2
show("X", X)
print("check 2X + A == 5B:", np.array_equal(2*X + A, 5*B))
```

## 8. יישום: תמונה דיגיטלית היא מטריצה

תמונה בגווני אפור היא מטריצה: כל איבר הוא עוצמת האור של פיקסל, בין 0 (שחור) ל-255 (לבן). ניקח שתי תמונות לדוגמה שמגיעות עם הספרייה `scikit-image` (חתול וכוס קפה), נהפוך אותן לגווני אפור ונביא אותן לאותו גודל – כדי שיהיה אפשר לחבר אותן.

</div>

```python
from skimage import data, color, transform

def gray(img, shape=(300, 450)):
    # RGB picture -> gray-level matrix with entries 0..255, resized to a common shape
    return transform.resize(color.rgb2gray(img), shape, anti_aliasing=True) * 255

cam = gray(data.chelsea())       # a cat
ast = gray(data.coffee())        # a cup of coffee

print("cat:", cam.shape, " pixel (0,0) =", round(cam[0, 0], 1), " min/max:", round(cam.min(), 1), round(cam.max(), 1))
print("a 5x5 corner of the matrix:")
print(cam[:5, :5])

def imshow(M, title="", ax=None):
    ax = ax or plt.gca()
    ax.imshow(np.clip(M, 0, 255), cmap="gray", vmin=0, vmax=255)
    ax.set_title(title); ax.axis("off")

fig, ax = plt.subplots(1, 2, figsize=(9, 4.5))
imshow(cam, "A (cat)", ax[0]); imshow(ast, "B (coffee)", ax[1]); plt.show()
```

### 8.1 הבהרה, נגטיב ושחלוף

- **הבהרה:** $A+40J$, כאשר $J$ היא מטריצה שכל איבריה 1 (בפייתון פשוט `A + 40`).
- **נגטיב:** $255J-A$.
- **שחלוף:** $A^{t}$ משקף את התמונה סביב האלכסון הראשי. שחלוף ואחר כך היפוך סדר העמודות נותן סיבוב ברבע סיבוב.

</div>

```python
J = np.ones_like(cam)
fig, ax = plt.subplots(1, 4, figsize=(16, 4.5))
imshow(cam, "A", ax[0])
imshow(cam + 40*J, "A + 40J  (brighter)", ax[1])
imshow(255*J - cam, "255J - A  (negative)", ax[2])
imshow(cam.T[:, ::-1], "A^t with columns reversed (rotation)", ax[3])
plt.show()
```

<div dir="rtl">

### 8.2 מעבר הדרגתי (cross-fade) בין שתי תמונות

$$F(t)=(1-t)\,A+t\,B,\qquad 0\le t\le 1$$

כל פריים בסרטון המעבר הוא **צירוף של שתי מטריצות** – חיבור וכפל בסקלר בלבד.

</div>

```python
ts = [0, 0.25, 0.5, 0.75, 1]
fig, ax = plt.subplots(1, len(ts), figsize=(18, 3.2))
for k, t in enumerate(ts):
    imshow((1 - t)*cam + t*ast, f"t = {t}", ax[k])
plt.show()
```

<div dir="rtl">

**הנפשה:** הריצו את התא כדי לראות את המעבר כסרטון קצר.

</div>

```python
from matplotlib import animation
from IPython.display import HTML

fig, ax = plt.subplots(figsize=(5, 3.5)); ax.axis("off")
im = ax.imshow(cam, cmap="gray", vmin=0, vmax=255)
def frame(k):
    t = k / 30
    im.set_data((1 - t)*cam + t*ast)
    ax.set_title(f"t = {t:.2f}")
    return [im]
anim = animation.FuncAnimation(fig, frame, frames=31, interval=80)
plt.close(fig)
HTML(anim.to_jshtml())
```

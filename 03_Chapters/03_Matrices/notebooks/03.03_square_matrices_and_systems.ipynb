## 1. חזקות של מטריצה (הגדרה 3.6.5, דוגמה 3.12)

</div>

```python
from numpy.linalg import matrix_power
A = np.array([[1, 2], [0, 1]])
for n in range(5):
    print(f"A^{n} =", matrix_power(A, n).tolist())
print("guess A^n = [[1, 2n], [0, 1]]  ->  A^10 =", matrix_power(A, 10).tolist())
```

```python
A = np.array([[0, 1], [-1, 0]])
for n in range(1, 9):
    print(f"A^{n} =", matrix_power(A, n).tolist())
print("A^2027 =", matrix_power(A, 2027).tolist())
```

## 2. יישום: מעבר לקוחות בין שני שירותי סטרימינג (שרשרת מרקוב)

$$x_{k+1}=Px_k,\qquad P=\begin{bmatrix}0.9&0.2\\0.1&0.8\end{bmatrix},\qquad x_k=P^kx_0$$

</div>

```python
P = np.array([[0.9, 0.2],
              [0.1, 0.8]])
x0 = np.array([500, 500])
xs = [x0]
for k in range(36):
    xs.append(P @ xs[-1])
xs = np.array(xs)
print("month 1:", xs[1], " month 2:", xs[2], " after a year:", xs[12].round(1))

plt.plot(xs[:, 0], label="service A"); plt.plot(xs[:, 1], label="service B")
plt.xlabel("month k"); plt.ylabel("customers (thousands)"); plt.legend(); plt.grid(True); plt.show()
print("P^36 =\n", matrix_power(P, 36).round(4))
```

## 3. מטריצות מתחלפות ומטריצות סקלריות (משפט 3.6.4)

</div>

```python
rng = np.random.default_rng(3)
A = rng.integers(-4, 5, (3, 3))
T = 7 * np.eye(3, dtype=int)                 # a scalar matrix
print("tI commutes with A:", np.array_equal(T @ A, A @ T))
D = np.diag([1, 2, 3])                        # diagonal but not scalar
print("diag(1,2,3) commutes with A:", np.array_equal(D @ A, A @ D))
```

## 5. משולשיות, אנטי-סימטריות ועקבה (לפי הסילבוס; לא בספר)

</div>

```python
rng = np.random.default_rng(4)
U1 = np.triu(rng.integers(-3, 4, (4, 4))); U2 = np.triu(rng.integers(-3, 4, (4, 4)))
show("U1 U2 (upper triangular)", U1 @ U2)
print("diagonal of U1U2:", np.diag(U1 @ U2), " = products of diagonals:", np.diag(U1) * np.diag(U2))
```

```python
A = np.array([[1, 4], [2, 3]])
S = (A + A.T) / 2; K = (A - A.T) / 2
show("symmetric part S", S); show("antisymmetric part K", K)
print("S + K == A:", np.allclose(S + K, A), "  K^t == -K:", np.allclose(K.T, -K))
print("tr(A) =", np.trace(A))
```

```python
A = rng.integers(-5, 6, (3, 3)); B = rng.integers(-5, 6, (3, 3))
print("tr(AB) =", np.trace(A @ B), "  tr(BA) =", np.trace(B @ A))
print("but AB == BA ?", np.array_equal(A @ B, B @ A))
print("tr(A)tr(B) =", np.trace(A) * np.trace(B), " (not equal to tr(AB) in general)")
```

## 6. יישום: מטריצת שכנות ומסלולים ברשת

$[A^k]_{ij}$ סופר את המסלולים באורך $k$ מ-$i$ ל-$j$.

</div>

```python
names = ["Dana", "Yossi", "Noa", "Avi", "Maya"]
edges = [(0, 1), (1, 2), (1, 3), (2, 3), (3, 4)]
n = len(names)
A = np.zeros((n, n), dtype=int)
for i, j in edges:
    A[i, j] = A[j, i] = 1
show("A (adjacency, symmetric)", A)
A2 = A @ A
show("A^2", A2)
for i in range(n):
    for j in range(i + 1, n):
        if A[i, j] == 0 and A2[i, j] > 0:
            print(f"suggest: {names[i]} - {names[j]}  ({A2[i, j]} common friend(s))")
```

## 7. מערכת בכתיב וקטורי ו-$\operatorname{Sol}(A|b)$ (דוגמה 3.15)

הספרייה `sympy` פותרת בדיוק (בשברים), ומחזירה את הפתרון הכללי עם פרמטרים.

</div>

```python
import sympy as sp
A = sp.Matrix([[1, 2, -1],
               [2, 4,  1]])
b = sp.Matrix([3, 9])
sol, params = A.gauss_jordan_solve(b)
print("Sol(A|b): x =", list(sol), " parameters:", list(params))
print("homogeneous solutions Sol(A|0): basis =", [list(v) for v in A.nullspace()])
print("check A*(4,0,1) =", list(A * sp.Matrix([4, 0, 1])))
```

**נסו בעצמכם:** פתרו את המערכת $3x-y=2,\ x+y+4z=0,\ 2y-z=7$ ובדקו אם $c=(1,1,-5)$ הוא פתרון.

</div>

```python
A = sp.Matrix([[3, -1, 0], [1, 1, 4], [0, 2, -1]]); b = sp.Matrix([2, 0, 7])
c = sp.Matrix([1, 1, -5])
print("A c =", list(A * c), " b =", list(b))
print("unique solution:", list(A.LUsolve(b)))
```
